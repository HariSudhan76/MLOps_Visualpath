## PROJECT 2 — Employee Salary Prediction (HR Analytics Domain) – Regression
##### 📌 Problem Statement:

An HR analytics team wants to predict the monthly salary of employees based on their demographic details, experience, performance score, department, job role, and company-related features.

Your job is to build an end-to-end ML regression pipeline that predicts Monthly Income for a new employee profile.

This is a real industry problem used in compensation analytics and HR automation.

📊 Get the dataset from Kaggle with attrition (target: salary(numerical))

📌 What you must do:

##### 1. Data Cleaning

    Remove employee_id
    Handle missing values (mean, median, or model-based imputation)
    Convert categorical columns (gender, overtime, education, department, job_role) using:
    OneHotEncoding or
    Ordinal encoding
    Convert Yes/No to 1/0
    Treat outliers (IQR or Winsorization)
    Fix data types

##### 2. Feature Engineering

    you must derive at least 2–3 new features, e.g.:
    experience_gap = years_of_experience - years_at_company
    overtime_flag = (overtime == "Yes")
    Salary per job level, job role grouping etc.

##### 3. Exploratory Data Analysis (EDA)

    Plot distribution of monthly income
    Income by job_role
    Income by department
    Income by job_level
    Correlation heatmap

##### 4. Train/Test Split

    80/20 split
    Random state fixed

##### 5. Model Training (At least 3 regression models)

    must train:
        Linear Regression
        RandomForestRegressor
        XGBRegressor

        Optional: SVR, CatBoost, LightGBM (if allowed)

##### 6. Model Evaluation

        you must compute:
            MAE (Mean Absolute Error)
            MSE (Mean Squared Error)
            RMSE (Root Mean Squared Error)
            R² Score
        Scatter plot: Predicted vs Actual

##### 7. Save Best Model
        Save final model as salary_prediction_model.pkl
        Save preprocessor (preprocessor.pkl)
